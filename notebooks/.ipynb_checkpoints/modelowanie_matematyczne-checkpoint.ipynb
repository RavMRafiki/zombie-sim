## Symulacja agentowa i metryki
Uruchomimy uproszczoną pętlę bez renderingu i zarejestrujemy metryki:
- Liczby `S` (ludzie+medycy+żołnierze), `I` (zarażeni+zombie), `R` (usunięci).
- Zliczymy przyrosty infekcji (lambda) i tempo leczenia/usuwania (gamma/mu).

```python
from main import upda
def count_groups(grid):
    z,h,i,m,s = grid.get_stats()
    S = h + m + s
    I = i + z
    # Usunięci: różnica względem początkowej populacji (przybliżenie)
    return S, I

def run_simulation(steps=500, seed=42, init=(38,45,0,10,25), training_mode=False):
    np.random.seed(seed)
    pygame.time.get_ticks()  # inicjalizacja timera
    grid = Grid(num_zombies=init[0], num_humans=init[1], num_infected=init[2], num_medics=init[3], num_soldiers=init[4])
    S_hist, I_hist, inf_events, heal_events = [], [], [], []
    prev_S, prev_I = None, None
    for t in range(steps):
        gm = grid.get_global_map_matrix()
        update_game_logic(grid, gm)
        S, I = count_groups(grid)
        S_hist.append(S); I_hist.append(I)
        if prev_S is not None and prev_I is not None:
            # Przyrost infekcji ~ spadek S gdy I>0
            dS = S - prev_S
            dI = I - prev_I
            # Zdarzenia (przybliżenia): ujemny dS -> infekcje, ujemny dI -> usunięci (zabici/leczeni)
            inf_events.append(max(0, -dS))
            heal_events.append(max(0, -dI))
        prev_S, prev_I = S, I
    return {"S": np.array(S_hist), "I": np.array(I_hist), "inf": np.array(inf_events), "heal": np.array(heal_events), "grid": grid}

res = run_simulation(steps=300)
print(f"Koniec symulacji. Ostatnie S={res['S'][-1]}, I={res['I'][-1]}")
plt.figure(figsize=(6,3))
plt.plot(res['S'], label='S (Susceptible)')
plt.plot(res['I'], label='I (Infectious)')
plt.legend(); plt.title('Dynamika agentowa'); plt.show()
```

## Estymacja parametrów SIR
Zaproponujemy estymację $\beta$ i $\gamma$ z danych symulacji:
- $\beta \approx \frac{\text{infekcje}}{(S\cdot I)/N}$,
- $\gamma \approx \frac{\text{usunięcia}}{I}$,
\ngdzie $N=S+I$ (pomijamy `R` jako stan usunięty).

```python
def estimate_params(S, I, inf, heal):
    S = S[:-1]; I = I[:-1]  # dopasuj długości
    N = S + I
    eps = 1e-9
    beta_t = inf / (np.maximum((S*I)/np.maximum(N, eps), eps))
    gamma_t = heal / np.maximum(I, eps)
    # Przytnij anomalie i uśrednij
    beta = np.nanmedian(np.clip(beta_t, 0, 1))
    gamma = np.nanmedian(np.clip(gamma_t, 0, 1))
    return float(beta), float(gamma)

beta, gamma = estimate_params(res['S'], res['I'], res['inf'], res['heal'])
print(f"Szacunek: beta={beta:.8f}, gamma={gamma:.8f}")
```

## Solver SIR (Euler) i porównanie
Rozwiążemy SIR w czasie dyskretnym metodą Eulera i porównamy z agentami.

```python
def sir_euler(S0, I0, beta, gamma, steps=300):
    S = np.zeros(steps); I = np.zeros(steps); R = np.zeros(steps)
    N = S0 + I0
    S[0], I[0], R[0] = S0, I0, 0
    for t in range(1, steps):
        dS = -beta * S[t-1] * I[t-1] / N
        dI = beta * S[t-1] * I[t-1] / N - gamma * I[t-1]
        dR = gamma * I[t-1]
        S[t] = max(0, S[t-1] + dS)
        I[t] = max(0, I[t-1] + dI)
        R[t] = max(0, R[t-1] + dR)
    return {"S":S, "I":I, "R":R}

sir = sir_euler(res['S'][0], res['I'][0], beta, gamma, steps=len(res['S']))
plt.figure(figsize=(6,3))
plt.plot(res['S'], label='Agent S', alpha=0.7)
plt.plot(res['I'], label='Agent I', alpha=0.7)
plt.plot(sir['S'], '--', label='SIR S')
plt.plot(sir['I'], '--', label='SIR I')
plt.legend(); plt.title('Porównanie agent vs SIR'); plt.show()
```

## Analiza kolejki (K2P_W02/W07)
Załóżmy M/M/c dla chorych oczekujących na medyka:
- `lambda`: średnia liczba nowych `Infected` na krok.
- `mu`: średnia liczba wyleczonych/usuniętych na krok per medyk.
- `c`: liczba medyków.

```python
def mmc_queue_metrics(lmbda, mu, c):
    rho = lmbda / (c*mu)
    # P0 (prawdopodobieństwo pustego systemu)
    def P0():
        s = sum([(lmbda/mu)**n / math.factorial(n) for n in range(c)])
        s += ((lmbda/mu)**c / math.factorial(c)) * (1/(1 - rho)) if rho < 1 else np.inf
        return 1/s
    P0v = P0()
    # Lq (średnia długość kolejki) (Formuła Erlanga-C)
    if rho >= 1:
        Lq = np.inf
    else:
        ErlangC = (((lmbda/mu)**c / math.factorial(c)) * (1/(1 - rho))) * P0v
        Lq = (ErlangC * rho) / (1 - rho)
    Wq = Lq / lmbda if lmbda > 0 else 0
    return {"rho":rho, "Lq":Lq, "Wq":Wq, "P0":P0v}

avg_inf = float(np.mean(res['inf'])) if len(res['inf'])>10 else 0.0
avg_heal = float(np.mean(res['heal'])) if len(res['heal'])>10 else 0.0
c_medics = sum(1 for ch in res['grid'].characters if isinstance(ch, Medic))
mu_per_medic = (avg_heal / max(c_medics,1))
q = mmc_queue_metrics(avg_inf, max(mu_per_medic,1e-6), max(c_medics,1))
print({"lambda":avg_inf, "mu":mu_per_medic, "c":c_medics, **q})
```

## Statystyka i Monte Carlo (K2P_W02)
Uruchomimy kilka replikacji, policzymy średnie i przedziały ufności dla `I` po T krokach.

```python
def monte_carlo_I(T=200, runs=10):
    finals = []
    for r in range(runs):
        resr = run_simulation(steps=T, seed=1234+r)
        finals.append(resr['I'][-1])
    arr = np.array(finals)
    mean = float(np.mean(arr))
    std = float(np.std(arr, ddof=1))
    # 95% CI dla średniej ~ t-approx przy n>=10
    ci = 1.96 * std / math.sqrt(len(arr))
    return {"mean":mean, "std":std, "ci95":ci, "samples":arr}

mc = monte_carlo_I(T=150, runs=12)
print({"mean_I": mc['mean'], "std": mc['std'], "ci95": mc['ci95']})
plt.figure(figsize=(5,3))
plt.hist(mc['samples'], bins=8, alpha=0.7)
plt.title('Rozkład I(T) w Monte Carlo'); plt.show()
```

## Podsumowanie (K2P_K02/K03)
- Agent-based model pokazuje lokalne interakcje i sygnały (krzyki, radio).
- SIR daje globalne przybliżenie z parametrami estymowanymi z danych.
- Analiza M/M/c przybliża obciążenie służby medycznej i czasy oczekiwania.
- Statystyka (średnie, przedziały ufności) ocenia stabilność modelu i wpływ parametrów.

Dalsze kroki: kalibracja `beta/gamma` przez dopasowanie (np. minimalizacja MSE),
eksperymenty z liczbą medyków/żołnierzy i oceną ryzyka przeciążenia.

```python
# Przełącznik uczenia dla przyspieszenia eksperymentów w notatniku
LEARNING_ENABLED = False

def update_game_logic(grid, global_map):
    # 1) FAZA DECYZYJNA — każdy liczy akcję na tej samej obserwacji
    decisions = []  # (char, current_state, action)
    for char in list(grid.characters):
        if isinstance(char, Infected):
            decisions.append((char, None, None))
            continue

        char.process_signals()

        visual_state = char.get_observation(global_map)
        vector_state = char.get_target_vector()
        current_state = (visual_state, vector_state)

        if isinstance(char, Zombie):
            action = zombie_agent.get_action(current_state)
        elif isinstance(char, Human) and not isinstance(char, Soldier):
            action = human_agent.get_action(current_state)
        elif isinstance(char, Soldier):
            action = soldier_agent.get_action(current_state)
        elif isinstance(char, Medic):
            action = char.get_autonomous_action()
        else:
            action = 4

        decisions.append((char, current_state, action))

    # 2) FAZA RUCHU — wykonujemy ruchy zgodnie z policzonymi akcjami
    move_results = {}  # char -> bool
    for char, _, action in decisions:
        if action is None:
            continue
        move_results[char] = char.move(action)

    # 3) FAZA AKCJI I NAGRÓD — liczymy act() i rewardy już po ruchach
    pending = []  # (agent, char, current_state, action, reward, done)

    for char, current_state, action in decisions:
        if isinstance(char, Infected):
            char.act()
            continue

        reward = 0.0
        done = False
        move_success = move_results.get(char, True)

        if not move_success:
            if isinstance(char, Zombie):
                reward -= 0.1
            elif isinstance(char, Human) and not isinstance(char, Soldier):
                reward -= 0.5
            elif isinstance(char, Soldier):
                reward -= 0.1

        if isinstance(char, Medic):
            char.act()
        elif isinstance(char, Zombie):
            if action == 4:
                if hasattr(char, 'prev_dist') and char.prev_dist < char.SIEGE_RANGE:
                    pass
                else:
                    reward -= 0.5
            reward += char.act()
            reward -= 0.05
            if not char.is_alive:
                reward -= 10
                done = True
        elif isinstance(char, Soldier):
            reward += char.act()
            if not char.is_alive:
                reward = -10
                done = True
        elif isinstance(char, Human):
            reward += char.act()
            if not char.is_alive:
                reward = -10
                done = True

        agent = None
        if isinstance(char, Soldier):
            agent = soldier_agent
        elif isinstance(char, Zombie):
            agent = zombie_agent
        elif isinstance(char, Human) and not isinstance(char, Soldier):
            agent = human_agent

        if agent is not None and current_state is not None:
            pending.append((agent, char, current_state, action, reward, done))

    # 4) NEXT STATE — jedna, spójna mapa po wszystkich ruchach i akcjach
    global_map_next = grid.get_global_map_matrix()

    for agent, char, current_state, action, reward, done in pending:
        new_visual = char.get_observation(global_map_next)
        new_vector = char.get_target_vector()
        next_state = (new_visual, new_vector)
        agent.memory.push(current_state, action, reward, next_state, done)
        if LEARNING_ENABLED:
            agent.learn()
```

## Proste metryki bez SIR
Poniżej liczymy łatwe do zrozumienia wskaźniki bez odwołania do modelu SIR:
- Nowe infekcje na krok + średnia krocząca.
- Łączna liczba infekcji (kumulacja).
- Szczyt `I` i czas do szczytu.
- Ostateczny udział narażonych, którzy uniknęli zakażenia (`S_end/N`).
- Przybliżony czas wygaszenia (pierwszy krok z `I=0`, jeśli wystąpi).
- Średnie tempo „usunięć” (leczenie/zabicie) na krok.

```python
import numpy as np
import matplotlib.pyplot as plt

# Upewnij się, że mamy wynik jednej symulacji 'res'
try:
    _ = res['S']
except Exception:
    res = run_simulation(steps=200)

S, I = res['S'], res['I']
N0 = float(S[0] + I[0])
new_inf = res['inf'] if len(res['inf'])>0 else np.zeros(max(len(S)-1,0))
removals = res['heal'] if len(res['heal'])>0 else np.zeros_like(new_inf)

# Średnia krocząca (okno 7)
def moving_avg(x, w=7):
    if len(x) < 1:
        return x
    w = min(w, max(1, len(x)))
    k = np.ones(w)/w
    return np.convolve(x, k, mode='same')

ma_inf = moving_avg(new_inf, 7)
cum_inf = np.cumsum(new_inf)

peak_I = int(np.max(I))
t_peak = int(np.argmax(I))
S_end = float(S[-1])
I_end = float(I[-1])
attack_final = 1.0 - S_end / N0  # udział kiedykolwiek zakażonych/"usuniętych"
time_to_ext = next((int(t) for t,v in enumerate(I) if v==0), None)
avg_removal = float(np.mean(removals)) if len(removals)>0 else 0.0

print({
    "peak_I": peak_I,
    "t_peak": t_peak,
    "S_end_share": round(S_end/N0,3),
    "attack_final": round(attack_final,3),
    "time_to_extinction": time_to_ext,
    "avg_new_infections_per_step": round(float(np.mean(new_inf)) if len(new_inf)>0 else 0.0, 3),
    "avg_removals_per_step": round(avg_removal,3)
})

fig, axs = plt.subplots(1,3, figsize=(12,3))
axs[0].plot(new_inf, alpha=0.5, label='Nowe infekcje/step')
axs[0].plot(ma_inf, label='Śr. krocząca (7)')
axs[0].legend(); axs[0].set_title('Nowe infekcje')

axs[1].plot(cum_inf, color='tab:orange')
axs[1].set_title('Kumulacja infekcji')

axs[2].plot(I, label='I')
axs[2].axvline(t_peak, ls='--', color='r', label=f't_peak={t_peak}')
axs[2].legend(); axs[2].set_title('I oraz czas do szczytu')
plt.tight_layout(); plt.show()
```

## Minimalna procentowa szansa roli zapewniająca wygaszenie w ≥80% uruchomień
Szukamy najmniejszej wartości `p` (udział roli przy rozmnażaniu ludzi), dla której w co najmniej 80% powtórzeń epidemia wygasa do zera w horyzoncie T.

```python
import random
from human import Human

def simulate_success(role_prob, steps=120, runs=6, seed_base=5000):
    Human.ROLE_CHANCE = float(role_prob)
    success = 0
    for r in range(runs):
        # Ustalony seed dla powtarzalności
        resr = run_simulation(steps=steps, seed=seed_base + r)
        I_series = resr['I']
        # Sukces: pojawia się krok z I==0 lub finalnie I==0
        extinct = any(int(v)==0 for v in I_series) or int(I_series[-1])==0
        success += 1 if extinct else 0
    return success / max(runs,1)

def find_min_role_percent(target=0.8, steps=120, runs=6, max_iter=12):
    lo, hi = 0.0, 1.0
    best = None
    history = []
    for _ in range(max_iter):
        mid = (lo + hi) / 2.0
        rate = simulate_success(mid, steps=steps, runs=runs)
        history.append((mid, rate))
        print(f"p={mid:.3f} -> success={rate:.2f}")
        if rate >= target:
            best = mid
            hi = mid
        else:
            lo = mid
    return best, history

best_p, hist = find_min_role_percent(target=0.8, steps=110, runs=6)
print({"best_role_percent": None if best_p is None else round(100*best_p,2)})
 
# Podgląd historii wyszukiwania
import matplotlib.pyplot as plt
xs = [p for p,_ in hist]
ys = [s for _,s in hist]
plt.figure(figsize=(5,3))
plt.plot(xs, ys, marker='o')
plt.axhline(0.8, ls='--', color='r', label='Cel 80%')
plt.xlabel('p (szansa roli)')
plt.ylabel('Odsetek wygaszeń')
plt.legend(); plt.title('Wyszukiwanie progu p'); plt.show()
```

## Próg udziału ról na starcie (medycy/żołnierze)
Wcześniej badaliśmy szansę roli przy rozmnażaniu, co ma słabszy wpływ. Teraz szukamy minimalnego udziału ról na starcie (odsetek wśród ludzi), który daje ≥80% wygaszeń w określonej liczbie powtórzeń.

```python
from human import Human
Human.ROLE_CHANCE = 0.0  # wyłącz losowanie ról przy rozmnażaniu, badamy tylko skład startowy

def simulate_success_initial_ratio(role_ratio, steps=80, runs=6, seed_base=8000, total_humans=45, base_zombies=38):
    medics = int(round(total_humans * role_ratio / 2))
    soldiers = int(round(total_humans * role_ratio - medics))
    humans  = max(0, total_humans - medics - soldiers)
    success = 0
    for r in range(runs):
        init = (base_zombies, humans, 0, medics, soldiers)
        resr = run_simulation(steps=steps, seed=seed_base + r, init=init)
        I_series = resr['I']
        extinct = (int(I_series[-1]) == 0)
        success += 1 if extinct else 0
    return success / max(runs,1)

def find_min_initial_ratio(target=0.8, steps=80, runs=6):
    # Prosty przeszukiwanie po siatce dla szybkości (0..1)
    grid = np.linspace(0.0, 1.0, 11)
    rates = []
    for p in grid:
        rate = simulate_success_initial_ratio(p, steps=steps, runs=runs)
        print(f"p={p:.2f} -> success={rate:.2f}")
        rates.append(rate)
    # Wybór najmniejszego p z wynikiem ≥ target
    best = None
    for p, rate in zip(grid, rates):
        if rate >= target:
            best = p
            break
    return best, grid, rates

best_init_p, grid_p, rates = find_min_initial_ratio(target=0.8, steps=60, runs=4)
print({"best_initial_percent": None if best_init_p is None else round(100*best_init_p,2)})

plt.figure(figsize=(5,3))
plt.plot(grid_p, rates, marker='o')
plt.axhline(0.8, ls='--', color='r', label='Cel 80%')
plt.xlabel('p (udział ról wśród ludzi)')
plt.ylabel('Odsetek wygaszeń (I==0 na końcu)')
plt.legend(); plt.title('Próg udziału ról na starcie'); plt.show()
```

## Próg "stłumienia" zamiast pełnego wygaszenia
Zamiast wymagać `I==0`, definiujemy sukces jako utrzymanie `I` poniżej małego progu `ε` przez ostatnie `W` kroków (realistyczne stłumienie epidemii).

```python
def is_suppressed(I_series, eps=2, tail=20):
    if len(I_series) < tail:
        tail = len(I_series)
    segment = I_series[-tail:]
    return int(np.max(segment)) <= eps

def simulate_success_initial_ratio_supp(role_ratio, steps=120, runs=8, seed_base=9000, total_humans=45, base_zombies=38, eps=2, tail=20):
    medics = int(round(total_humans * role_ratio / 2))
    soldiers = int(round(total_humans * role_ratio - medics))
    humans  = max(0, total_humans - medics - soldiers)
    success = 0
    for r in range(runs):
        init = (base_zombies, humans, 0, medics, soldiers)
        resr = run_simulation(steps=steps, seed=seed_base + r, init=init)
        if is_suppressed(resr['I'], eps=eps, tail=tail):
            success += 1
    return success / max(runs,1)

def find_min_initial_ratio_supp(target=0.8, steps=120, runs=8, eps=2, tail=20):
    grid = np.linspace(0.0, 1.0, 11)
    rates = []
    for p in grid:
        rate = simulate_success_initial_ratio_supp(p, steps=steps, runs=runs, eps=eps, tail=tail)
        print(f"p={p:.2f} -> suppressed={rate:.2f}")
        rates.append(rate)
    best = None
    for p, rate in zip(grid, rates):
        if rate >= target:
            best = p
            break
    return best, grid, rates

best_init_p_supp, grid_p_s, rates_s = find_min_initial_ratio_supp(target=0.8, steps=90, runs=6, eps=3, tail=25)
print({"best_initial_percent_suppression": None if best_init_p_supp is None else round(100*best_init_p_supp,2)})

plt.figure(figsize=(5,3))
plt.plot(grid_p_s, rates_s, marker='o')
plt.axhline(0.8, ls='--', color='r', label='Cel 80%')
plt.xlabel('p (udział ról wśród ludzi)')
plt.ylabel('Odsetek stłumień (I≤ε przez ostatnie W kroków)')
plt.legend(); plt.title('Próg stłumienia (ε, W)'); plt.show()
```
